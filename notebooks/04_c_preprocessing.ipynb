```python
import re

def flag_suspicious_skill(skill):
    skill = str(skill).strip()

    reasons = []

    # 지나치게 긴 문자열
    if len(skill) >= 40:
        reasons.append("too_long")

    # 문장형 한국어 종결 표현
    sentence_endings = [
        "합니다",
        "해요",
        "됩니다",
        "있어요",
        "있습니다",
        "주세요",
        "합니다.",
        "해요.",
        "됩니다.",
        "있어요.",
        "있습니다.",
    ]

    if any(
        ending in skill
        for ending in sentence_endings
    ):
        reasons.append("sentence_like")

    # 마침표가 포함된 긴 문장
    if "." in skill and len(skill) >= 20:
        reasons.append("sentence_punctuation")

    # 공백이 지나치게 많은 경우
    if skill.count(" ") >= 6:
        reasons.append("too_many_words")

    return " | ".join(reasons)


skill_dictionary_df["review_reason"] = (
    skill_dictionary_df["skill"]
    .apply(flag_suspicious_skill)
)

suspicious_skills_df = (
    skill_dictionary_df[
        skill_dictionary_df["review_reason"] != ""
    ]
    .sort_values(
        ["count", "skill"],
        ascending=[True, True]
    )
    .reset_index(drop=True)
)

print(
    "suspicious skill candidates:",
    len(suspicious_skills_df)
)

display(
    suspicious_skills_df
)
```

```python
# --------------------------------------------------
# 이상 Skill의 원본 공고 추적
# --------------------------------------------------

suspicious_skill_names = set(
    suspicious_skills_df["skill"]
)

suspicious_rows = []

for _, row in jobs_df.iterrows():
    skills = [
        skill.strip()
        for skill in str(
            row["skills_std"]
        ).split("|")
        if skill.strip()
    ]

    matched = [
        skill
        for skill in skills
        if skill in suspicious_skill_names
    ]

    if matched:
        suspicious_rows.append({
            "job_id": row["job_id"],
            "company": row["company"],
            "title": row["title"],
            "matched_skills": " | ".join(
                matched
            ),
            "skills_raw": row["skills_raw"],
            "skills_std": row["skills_std"],
        })

suspicious_rows_df = pd.DataFrame(
    suspicious_rows
)

print(
    "affected job rows:",
    len(suspicious_rows_df)
)

display(
    suspicious_rows_df
)
```

```python
# --------------------------------------------------
# 문장형 오염 Skill 제거
# --------------------------------------------------

def is_valid_skill(skill):
    skill = str(skill).strip()

    if not skill:
        return False

    # 문장형 한국어 표현 제거
    sentence_patterns = [
        "해요",
        "합니다",
        "됩니다",
        "있어요",
        "있습니다",
        "주세요",
        "나누고",
        "오시면",
    ]

    if any(
        pattern in skill
        for pattern in sentence_patterns
    ):
        return False

    # 지나치게 긴 항목 제거
    if len(skill) >= 40:
        return False

    return True


def clean_skill_string(value):
    if pd.isna(value):
        return ""

    skills = [
        skill.strip()
        for skill in str(value).split("|")
        if skill.strip()
    ]

    valid_skills = [
        skill
        for skill in skills
        if is_valid_skill(skill)
    ]

    # 순서 유지 + 중복 제거
    valid_skills = list(
        dict.fromkeys(valid_skills)
    )

    return " | ".join(valid_skills)


jobs_df["skills_std_clean"] = (
    jobs_df["skills_std"]
    .apply(clean_skill_string)
)
```

```python
# --------------------------------------------------
# 문장형 오염 Skill 제거
# --------------------------------------------------

def is_valid_skill(skill):
    skill = str(skill).strip()

    if not skill:
        return False

    # 1. 명백한 문장형/설명형 표현 제거
    sentence_patterns = [
        "해요",
        "합니다",
        "됩니다",
        "있어요",
        "있습니다",
        "주세요",
        "나누고",
        "오시면",
        "이라면",
        "아니더라도",
        "성장은 물론",
        "관심이 많은",
        "통해 버그를",
        "개발 노하우에 대해",
        "엔지니어링 데이",
        "Tech Talk 진행",
    ]

    if any(
        pattern in skill
        for pattern in sentence_patterns
    ):
        return False

    # 2. 지나치게 긴 문자열 제거
    if len(skill) >= 40:
        return False

    # 3. 너무 긴 한국어 설명형 구절 제거
    korean_chars = sum(
        1
        for ch in skill
        if "가" <= ch <= "힣"
    )

    if korean_chars >= 12 and skill.count(" ") >= 3:
        return False

    return True


def clean_skill_string(value):
    if pd.isna(value):
        return ""

    skills = [
        skill.strip()
        for skill in str(value).split("|")
        if skill.strip()
    ]

    valid_skills = [
        skill
        for skill in skills
        if is_valid_skill(skill)
    ]

    # 순서 유지 + 중복 제거
    valid_skills = list(
        dict.fromkeys(valid_skills)
    )

    return " | ".join(valid_skills)


jobs_df["skills_std_clean"] = (
    jobs_df["skills_std"]
    .apply(clean_skill_string)
)
```

```python
# --------------------------------------------------
# Skill Category 2차 자동 분류
# --------------------------------------------------

def classify_skill_category(skill):
    skill = str(skill).strip()
    lower = skill.lower()

    # 이미 1차 매핑된 기술은 그대로 사용
    if skill in SKILL_CATEGORY_MAP:
        return SKILL_CATEGORY_MAP[skill]

    # Programming Language
    programming_languages = {
        "scala", "r", "dart", "swift",
        "objective-c", "php", "ruby",
        "shell", "bash", "groovy",
    }

    if lower in programming_languages:
        return "Programming Language"

    # Backend / Framework
    backend_keywords = [
        "spring",
        "nestjs",
        "node.js",
        "nodejs",
        "django",
        "fastapi",
        "flask",
        "grpc",
        "graphql",
        "hibernate",
        "mybatis",
        "batch",
    ]

    if any(
        keyword in lower
        for keyword in backend_keywords
    ):
        return "Backend Framework"

    # Frontend / Web
    frontend_keywords = [
        "react",
        "vue",
        "angular",
        "next.js",
        "webpack",
        "vite",
        "babel",
        "swc",
        "rspack",
        "rolldown",
        "stylex",
        "emotion",
        "recoil",
        "jotai",
        "zustand",
        "vanilla-extract",
        "react-query",
        "yarn",
        "pnpm",
    ]

    if any(
        keyword in lower
        for keyword in frontend_keywords
    ):
        return "Frontend"

    # Database / Search / Storage
    database_keywords = [
        "mysql",
        "postgres",
        "oracle",
        "mongodb",
        "redis",
        "elasticsearch",
        "opensearch",
        "dynamodb",
        "cassandra",
        "mariadb",
        "clickhouse",
        "database",
    ]

    if any(
        keyword in lower
        for keyword in database_keywords
    ):
        return "Database"

    # Data Engineering / Analytics
    data_keywords = [
        "spark",
        "hadoop",
        "airflow",
        "dbt",
        "impala",
        "flink",
        "trino",
        "presto",
        "databricks",
        "snowflake",
        "bigquery",
        "etl",
        "data pipeline",
    ]

    if any(
        keyword in lower
        for keyword in data_keywords
    ):
        return "Data Engineering"

    # AI / ML
    ai_keywords = [
        "llm",
        "tensorflow",
        "pytorch",
        "scikit",
        "sklearn",
        "xgboost",
        "lightgbm",
        "hugging face",
        "huggingface",
        "transformer",
        "bert",
        "retrieval",
        "embedding",
        "tensorrt",
        "cuda",
        "onnx",
        "langchain",
        "claude",
        "openai",
        "machine learning",
        "deep learning",
    ]

    # RAG는 짧은 키워드라 부분 문자열이 아니라
    # 기술명이 정확히 RAG일 때만 분류
    if lower == "rag":
        return "AI/ML"

    if any(
        keyword in lower
        for keyword in ai_keywords
    ):
        return "AI/ML"

    # Cloud
    cloud_keywords = [
        "aws",
        "gcp",
        "google cloud",
        "azure",
        "cloudflare",
        "cloudfront",
        "s3",
        "ec2",
        "lambda",
    ]

    if any(
        keyword in lower
        for keyword in cloud_keywords
    ):
        return "Cloud"

    # DevOps / Infrastructure
    devops_keywords = [
        "docker",
        "kubernetes",
        "terraform",
        "jenkins",
        "argocd",
        "gocd",
        "github actions",
        "helm",
        "istio",
        "vault",
        "harbor",
        "linux",
        "nginx",
        "haproxy",
        "ci/cd",
    ]

    if any(
        keyword in lower
        for keyword in devops_keywords
    ):
        return "DevOps/Cloud"

    # Observability / Monitoring
    observability_keywords = [
        "grafana",
        "prometheus",
        "thanos",
        "elk",
        "kibana",
        "logstash",
        "datadog",
        "sentry",
        "monitoring",
        "observability",
    ]

    if any(
        keyword in lower
        for keyword in observability_keywords
    ):
        return "Observability"

    # Messaging
    messaging_keywords = [
        "kafka",
        "rabbitmq",
        "sqs",
        "pub/sub",
        "pubsub",
        "messaging",
    ]

    if any(
        keyword in lower
        for keyword in messaging_keywords
    ):
        return "Data/Messaging"

    # Build / Package Tool
    build_keywords = [
        "gradle",
        "maven",
        "npm",
    ]

    if any(
        keyword == lower
        for keyword in build_keywords
    ):
        return "Build/Package Tool"

    # Security
    security_keywords = [
        "security",
        "iam",
        "oauth",
        "jwt",
        "owasp",
        "waf",
        "siem",
        "zero trust",
    ]

    if any(
        keyword in lower
        for keyword in security_keywords
    ):
        return "Security"

    # Collaboration / Development Tool
    collaboration_keywords = [
        "git",
        "github",
        "gitlab",
        "jira",
        "confluence",
        "slack",
    ]

    if any(
        keyword in lower
        for keyword in collaboration_keywords
    ):
        return "Collaboration"

    return "Unclassified"


skill_dictionary_final_df["category"] = (
    skill_dictionary_final_df["skill"]
    .apply(classify_skill_category)
)


unclassified_skills_df = (
    skill_dictionary_final_df[
        skill_dictionary_final_df[
            "category"
        ] == "Unclassified"
    ]
    .reset_index(drop=True)
)

print(
    "unclassified after second pass:",
    len(unclassified_skills_df)
)

display(
    unclassified_skills_df
)
```

```python
# --------------------------------------------------
# Skill Category 3차 자동 분류
# --------------------------------------------------

def classify_skill_category_v3(skill):
    current = classify_skill_category(skill)

    if current != "Unclassified":
        return current

    skill = str(skill).strip()
    lower = skill.lower()

    # AI / ML / Compute
    ai_compute_keywords = [
        "gpu",
        "genai",
        "generative ai",
        "cuda",
        "tensorrt",
        "inference",
        "model serving",
        "vector db",
        "vector database",
        "feature store",
    ]

    if any(
        keyword in lower
        for keyword in ai_compute_keywords
    ):
        return "AI/ML"

    # Data Engineering / Lakehouse
    data_platform_keywords = [
        "iceberg",
        "hudi",
        "delta lake",
        "lakehouse",
        "data lake",
        "warehouse",
        "data warehouse",
        "olap",
        "oltp",
    ]

    if any(
        keyword in lower
        for keyword in data_platform_keywords
    ):
        return "Data Engineering"

    # Runtime / Platform
    runtime_keywords = [
        "jvm",
        "runtime",
        "jre",
        "jdk",
    ]

    if any(
        keyword in lower
        for keyword in runtime_keywords
    ):
        return "Runtime/Platform"

    # Storage / Infrastructure
    storage_keywords = [
        "ceph",
        "storage",
        "object storage",
        "block storage",
        "filesystem",
        "file system",
    ]

    if any(
        keyword in lower
        for keyword in storage_keywords
    ):
        return "Storage/Infrastructure"

    # Network / Protocol
    network_keywords = [
        "http",
        "https",
        "tcp",
        "udp",
        "websocket",
        "web socket",
        "grpc",
        "dns",
        "network",
        "protocol",
        "rest api",
        "rest",
    ]

    if any(
        keyword in lower
        for keyword in network_keywords
    ):
        return "Network/Protocol"

    # Security
    security_keywords_v3 = [
        "firewall",
        "ids",
        "ips",
        "vulnerability",
        "penetration",
        "threat",
        "malware",
        "encryption",
        "cryptography",
        "authentication",
        "authorization",
    ]

    if any(
        keyword in lower
        for keyword in security_keywords_v3
    ):
        return "Security"

    # Frontend
    frontend_keywords_v3 = [
        "tanstack query",
        "tanstack",
        "storybook",
        "tailwind",
        "styled-components",
        "css",
        "html",
        "frontend",
        "webview",
    ]

    if any(
        keyword in lower
        for keyword in frontend_keywords_v3
    ):
        return "Frontend"

    # Mobile
    mobile_keywords = [
        "android",
        "ios",
        "flutter",
        "react native",
        "swiftui",
        "jetpack compose",
    ]

    if any(
        keyword in lower
        for keyword in mobile_keywords
    ):
        return "Mobile"

    # Architecture / General Engineering
    architecture_keywords = [
        "microservice",
        "microservices",
        "msa",
        "ddd",
        "domain-driven",
        "event driven",
        "event-driven",
        "architecture",
        "distributed system",
        "distributed systems",
    ]

    if any(
        keyword in lower
        for keyword in architecture_keywords
    ):
        return "Architecture"

    return "Unclassified"


skill_dictionary_final_df["category"] = (
    skill_dictionary_final_df["skill"]
    .apply(classify_skill_category_v3)
)


unclassified_v3_df = (
    skill_dictionary_final_df[
        skill_dictionary_final_df["category"]
        == "Unclassified"
    ]
    .reset_index(drop=True)
)

print(
    "unclassified after third pass:",
    len(unclassified_v3_df)
)

display(unclassified_v3_df)
```

```python
# --------------------------------------------------
# Vector 문맥별 의미 분리
# --------------------------------------------------

VECTOR_SEARCH_JOB_IDS = {
    "toss_7646941003",
    "toss_5498959003",
}

VECTOR_TOOL_JOB_IDS = {
    "toss_8006356003",
}


def replace_skill_token(value, old_skill, new_skill):
    if pd.isna(value):
        return value

    skills = [
        skill.strip()
        for skill in str(value).split("|")
        if skill.strip()
    ]

    skills = [
        new_skill if skill == old_skill else skill
        for skill in skills
    ]

    return " | ".join(
        dict.fromkeys(skills)
    )


# AI/ML 공고:
# Vector -> Vector Search
mask_vector_search = (
    jobs_df["job_id"]
    .isin(VECTOR_SEARCH_JOB_IDS)
)

jobs_df.loc[
    mask_vector_search,
    "skills_std_clean"
] = (
    jobs_df.loc[
        mask_vector_search,
        "skills_std_clean"
    ]
    .apply(
        lambda x: replace_skill_token(
            x,
            "Vector",
            "Vector Search",
        )
    )
)


# DataOps 공고:
# Vector 로그 수집 도구이므로 이름 명확화
mask_vector_tool = (
    jobs_df["job_id"]
    .isin(VECTOR_TOOL_JOB_IDS)
)

jobs_df.loc[
    mask_vector_tool,
    "skills_std_clean"
] = (
    jobs_df.loc[
        mask_vector_tool,
        "skills_std_clean"
    ]
    .apply(
        lambda x: replace_skill_token(
            x,
            "Vector",
            "Vector (Observability)",
        )
    )
)
```
