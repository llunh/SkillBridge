```python
def map_toss_job_category(job):
    raw_category = get_metadata_value(
        job,
        "커리어 페이지 노출 Job Category 값을 선택해주세요"
    )

    title = job.get("title", "").lower()

    # --------------------------------------------------
    # 명확한 카테고리
    # --------------------------------------------------
    if raw_category == "Backend":
        return "Backend"

    if raw_category == "Frontend":
        return "Frontend"

    if raw_category == "ML":
        return "AI/ML"

    if raw_category in {
        "Data Engineering",
        "Data Analysis",
    }:
        return "Data"

    if raw_category == "Security Engineering":
        return "Security"

    if raw_category == "App":
        return "Mobile"

    # --------------------------------------------------
    # Infra는 제목 기반으로 세분화
    # --------------------------------------------------
    if raw_category == "Infra":

        # Security
        if (
            "security" in title
            or "보안" in title
        ):
            return "Security"

        # DevOps / Cloud / Infra Engineering
        if any(
            keyword in title
            for keyword in [
                "engineer",
                "엔지니어",
                "aiops",
                "systems",
                "system",
                "network",
                "infrastructure operations",
                "콜 인프라 엔지니어",
            ]
        ):
            return "DevOps/Cloud"

        return "Other"

    # --------------------------------------------------
    # 분석 범위에서 제외
    # --------------------------------------------------
    if raw_category in {
        "Information Security",
        "QA",
        "AML",
        "Data Managing",
    }:
        return "Other"

    return "Other"
```

```python
def classify_toss_heading(heading):
    h = heading.strip()

    # 업무
    if (
        "함께할 업무" in h
        or "함께 할 업무" in h
    ):
        return "responsibilities"

    # 필수요건
    if (
        "이런 분과 함께하고 싶어요" in h
        or "이런 분을 기다리고" in h
    ):
        return "requirements"

    # 우대사항
    if (
        "더 좋아요" in h
        or "더 좋습니다" in h
    ):
        return "preferred"

    # 팀 소개
    if "합류하게 될 팀" in h:
        return "team_intro"

    # 기술 스택
    if (
        "사용하는 기술" in h
        or "기술을 사용하고" in h
    ):
        return "tech_stack"

    # 이력서 작성 가이드
    if (
        "이력서" in h
        or "지원서" in h
    ):
        return "resume_guide"

    # 채용 절차
    if (
        "합류여정" in h
        or "합류 여정" in h
    ):
        return "process"

    # 안내사항
    if (
        "꼭 확인" in h
        or "참고해" in h
    ):
        return "notice"

    return "other"


heading_type_counter = Counter()

other_headings = Counter()

for heading, count in heading_counter.items():
    heading_type = classify_toss_heading(heading)

    heading_type_counter[heading_type] += count

    if heading_type == "other":
        other_headings[heading] += count


print("===== Toss Heading 의미 분류 =====")

for heading_type, count in heading_type_counter.most_common():
    print(
        f"{heading_type}: {count}"
    )

print("\n===== 아직 분류되지 않은 Heading =====")

for heading, count in other_headings.most_common():
    print(
        f"{count:3} | {heading}"
    )
```

```python
def classify_toss_heading(heading):
    h = heading.strip()

    # --------------------------------------------------
    # 업무
    # --------------------------------------------------
    if (
        "함께할 업무" in h
        or "함께 할 업무" in h
        or "함께 할 일이에요" in h
        or "함께할 일이에요" in h
    ):
        return "responsibilities"

    # --------------------------------------------------
    # 필수요건
    # --------------------------------------------------
    if (
        "이런 분과 함께하고 싶어요" in h
        or "이런 분을 기다리고" in h
    ):
        return "requirements"

    # --------------------------------------------------
    # 우대사항
    # --------------------------------------------------
    if (
        "더 좋아요" in h
        or "더 좋습니다" in h
        or "더욱 좋아요" in h
        or "더욱 좋습니다" in h
    ):
        return "preferred"

    # --------------------------------------------------
    # 팀 / 조직 소개
    # --------------------------------------------------
    if (
        "합류하게 될 팀" in h
        or "합류하게 될 도메인" in h
        or "조직 구조" in h
        or "Division" in h
        or "소개해드려요" in h
        or "소개해 드려요" in h
    ):
        return "team_intro"

    # --------------------------------------------------
    # 기술 스택
    # --------------------------------------------------
    if (
        "사용하는 기술" in h
        or "기술을 사용하고" in h
    ):
        return "tech_stack"

    # --------------------------------------------------
    # 이력서 / 지원서 작성 가이드
    # --------------------------------------------------
    if (
        "이력서" in h
        or "지원서" in h
    ):
        return "resume_guide"

    # --------------------------------------------------
    # 채용 절차
    # --------------------------------------------------
    if (
        "합류여정" in h
        or "합류 여정" in h
        or "합류 과정" in h
    ):
        return "process"

    # --------------------------------------------------
    # 안내 / 확인사항
    # --------------------------------------------------
    if (
        "꼭 확인" in h
        or "반드시 확인" in h
        or "참고해" in h
    ):
        return "notice"

    return "other"


heading_type_counter = Counter()
other_headings = Counter()

for heading, count in heading_counter.items():
    heading_type = classify_toss_heading(heading)

    heading_type_counter[heading_type] += count

    if heading_type == "other":
        other_headings[heading] += count


print("===== Toss Heading 의미 분류 v2 =====")

for heading_type, count in heading_type_counter.most_common():
    print(
        f"{heading_type}: {count}"
    )

print("\n===== 아직 분류되지 않은 Heading v2 =====")

for heading, count in other_headings.most_common():
    print(
        f"{count:3} | {heading}"
    )
```

```python
heading_level_counter = Counter()
heading_level_samples = {
    1: [],
    2: [],
    3: [],
    4: [],
}

for job in target_toss_jobs:
    description = get_toss_job_description(job)

    matches = re.findall(
        r"(?m)^\s*(#{1,4})\s*(.*?)\s*$",
        description
    )

    for hashes, heading in matches:
        level = len(hashes)

        heading_level_counter[level] += 1

        if (
            heading.strip()
            and len(heading_level_samples[level]) < 15
        ):
            heading_level_samples[level].append(
                heading.strip()
            )


print("===== Toss Heading Level 분포 =====")

for level, count in sorted(
    heading_level_counter.items()
):
    print(
        f"H{level}: {count}"
    )


print("\n===== Level별 샘플 =====")

for level in sorted(
    heading_level_samples
):
    print(
        f"\n[H{level}]"
    )

    for heading in heading_level_samples[level]:
        print(
            "-",
            heading
        )
```

```python
def parse_toss_description(description):
    if not description:
        return {
            "team_intro": "",
            "responsibilities": "",
            "requirements": "",
            "preferred": "",
            "tech_stack": "",
            "resume_guide": "",
            "process": "",
            "notice": "",
            "other": "",
        }

    # H1(#)만 섹션 경계로 사용
    pattern = re.compile(
        r"(?m)^\s*#(?!#)\s*(.+?)\s*$"
    )

    matches = list(pattern.finditer(description))

    result = {
        "team_intro": [],
        "responsibilities": [],
        "requirements": [],
        "preferred": [],
        "tech_stack": [],
        "resume_guide": [],
        "process": [],
        "notice": [],
        "other": [],
    }

    # 첫 H1 이전 텍스트가 있다면 other에 보존
    if matches:
        prefix = description[:matches[0].start()].strip()

        if prefix:
            result["other"].append(prefix)

    for i, match in enumerate(matches):
        heading = match.group(1).strip()

        start = match.end()

        if i + 1 < len(matches):
            end = matches[i + 1].start()
        else:
            end = len(description)

        body = description[start:end].strip()

        section_type = classify_toss_heading(
            heading
        )

        if body:
            result[section_type].append(body)

    # 리스트 → 문자열
    for key in result:
        result[key] = "\n".join(
            result[key]
        ).strip()

    return result


# 첫 번째 Android 공고로 테스트
android_description = get_toss_job_description(
    target_toss_jobs[0]
)

android_parsed = parse_toss_description(
    android_description
)

for key, value in android_parsed.items():
    print("\n" + "=" * 80)
    print(f"[{key}]")
    print(value[:1200])
```

```python
def classify_toss_heading(heading):
    h = heading.strip()

    # --------------------------------------------------
    # 업무
    # --------------------------------------------------
    if (
        "함께할 업무" in h
        or "함께 할 업무" in h
        or "함께 할 일이에요" in h
        or "함께할 일이에요" in h
        or h == "합류하면 함께 할 일"
        or h == "합류하면 함께할 일"
    ):
        return "responsibilities"

    # --------------------------------------------------
    # 필수요건
    # --------------------------------------------------
    if (
        "이런 분과 함께하고 싶어요" in h
        or "이런분과 함께하고 싶어요" in h
        or "이런 분을 기다리고" in h
        or "역량이 꼭 필요해요" in h
    ):
        return "requirements"

    # --------------------------------------------------
    # 우대사항
    # --------------------------------------------------
    if (
        "더 좋아요" in h
        or "더 좋습니다" in h
        or "더욱 좋아요" in h
        or "더욱 좋습니다" in h
    ):
        return "preferred"

    # --------------------------------------------------
    # 팀 / 조직 소개
    # --------------------------------------------------
    if (
        "합류하게 될 팀" in h
        or "함께하게 될 팀" in h
        or "합류하게 될 도메인" in h
        or "조직 구조" in h
        or "Division" in h
        or "소개해드려요" in h
        or "소개해 드려요" in h
    ):
        return "team_intro"

    # --------------------------------------------------
    # 기술 스택
    # --------------------------------------------------
    if (
        "사용하는 기술" in h
        or "기술을 사용하고" in h
        or "이런 기술을 사용" in h
    ):
        return "tech_stack"

    # --------------------------------------------------
    # 이력서 / 경력기술서
    # --------------------------------------------------
    if (
        "이력서" in h
        or "지원서" in h
        or "경력기술서" in h
    ):
        return "resume_guide"

    # --------------------------------------------------
    # 채용 절차
    # --------------------------------------------------
    if (
        "합류여정" in h
        or "합류 여정" in h
        or "합류 과정" in h
    ):
        return "process"

    # --------------------------------------------------
    # 안내사항
    # --------------------------------------------------
    if (
        "꼭 확인" in h
        or "반드시 확인" in h
        or "참고해" in h
    ):
        return "notice"

    # --------------------------------------------------
    # 동료 인터뷰 / 코멘트
    # --------------------------------------------------
    if (
        "동료를 위한 한마디" in h
        or "동료의 한마디" in h
    ):
        return "colleague_note"

    return "other"


def parse_toss_description(description):
    if not description:
        return {
            "team_intro": "",
            "responsibilities": "",
            "requirements": "",
            "preferred": "",
            "tech_stack": "",
            "resume_guide": "",
            "process": "",
            "notice": "",
            "colleague_note": "",
            "other": "",
        }

    pattern = re.compile(
        r"(?m)^\s*#(?!#)\s*(.+?)\s*$"
    )

    matches = list(pattern.finditer(description))

    result = {
        "team_intro": [],
        "responsibilities": [],
        "requirements": [],
        "preferred": [],
        "tech_stack": [],
        "resume_guide": [],
        "process": [],
        "notice": [],
        "colleague_note": [],
        "other": [],
    }

    core_sections = {
        "responsibilities",
        "requirements",
        "preferred",
        "tech_stack",
    }

    current_section = None

    if matches:
        prefix = description[
            :matches[0].start()
        ].strip()

        if prefix:
            result["other"].append(prefix)

    for i, match in enumerate(matches):
        heading = match.group(1).strip()

        start = match.end()

        if i + 1 < len(matches):
            end = matches[i + 1].start()
        else:
            end = len(description)

        body = description[start:end].strip()

        section_type = classify_toss_heading(
            heading
        )

        # 명확하게 분류되는 heading
        if section_type != "other":
            current_section = section_type

            if body:
                result[section_type].append(body)

            continue

        # 분류되지 않은 H1이지만 현재 핵심 섹션 안이라면
        # 소제목으로 간주하여 현재 섹션에 보존
        if current_section in core_sections:
            subsection_text = heading

            if body:
                subsection_text += "\n" + body

            result[current_section].append(
                subsection_text
            )

        else:
            other_text = heading

            if body:
                other_text += "\n" + body

            result["other"].append(
                other_text
            )

    for key in result:
        result[key] = "\n".join(
            result[key]
        ).strip()

    return result
```

```python
experience_pattern = re.compile(
    r"(?:"
    r"신입(?:/경력)?"
    r"|경력(?:직)?"
    r"|\d+\s*년\s*(?:이상|이하|미만|초과)"
    r"|최소\s*\d+\s*년"
    r")"
)


def extract_toss_experience(job):
    description = get_toss_job_description(job)

    parsed = parse_toss_description(description)
    parsed = apply_toss_fallbacks(
        job,
        parsed
    )

    title = job.get("title", "")
    requirements = parsed["requirements"]

    # 1순위: 공고 제목
    title_matches = experience_pattern.findall(
        title
    )

    if title_matches:
        return " | ".join(
            dict.fromkeys(title_matches)
        )

    # 2순위: 지원자격
    requirement_matches = experience_pattern.findall(
        requirements
    )

    if requirement_matches:
        return " | ".join(
            dict.fromkeys(requirement_matches)
        )

    # 명시적 경력 조건 없음
    return ""


experience_test = pd.DataFrame([
    {
        "id": job.get("id"),
        "title": job.get("title"),
        "experience": extract_toss_experience(job),
    }
    for job in target_toss_jobs
])

print("===== Toss experience 추출 결과 =====")
print(
    "experience 있음:",
    (experience_test["experience"] != "").sum(),
    "/",
    len(experience_test)
)

display(
    experience_test[
        experience_test["experience"] != ""
    ].head(30)
)
```

```python
numeric_experience_pattern = re.compile(
    r"(?:"
    r"\d+\s*년\s*(?:이상|이하|미만|초과)"
    r"|최소\s*\d+\s*년"
    r")"
)

general_experience_pattern = re.compile(
    r"(?:신입/경력|신입|경력직|경력)"
)


def extract_toss_experience(job):
    description = get_toss_job_description(job)

    parsed = parse_toss_description(description)
    parsed = apply_toss_fallbacks(
        job,
        parsed
    )

    title = job.get("title", "")
    requirements = parsed["requirements"]

    # --------------------------------
    # 1순위: 제목의 숫자 경력 조건
    # --------------------------------
    title_numeric = numeric_experience_pattern.findall(
        title
    )

    if title_numeric:
        return " | ".join(
            dict.fromkeys(title_numeric)
        )

    # --------------------------------
    # 2순위: 제목의 신입/경력 표현
    # --------------------------------
    title_general = general_experience_pattern.findall(
        title
    )

    if title_general:
        return " | ".join(
            dict.fromkeys(title_general)
        )

    # --------------------------------
    # 3순위: 지원자격의 숫자 경력 조건
    # --------------------------------
    requirement_numeric = numeric_experience_pattern.findall(
        requirements
    )

    if requirement_numeric:
        return " | ".join(
            dict.fromkeys(requirement_numeric)
        )

    # --------------------------------
    # 4순위: 지원자격의 일반 경력 표현
    # --------------------------------
    requirement_general = general_experience_pattern.findall(
        requirements
    )

    if requirement_general:
        return " | ".join(
            dict.fromkeys(requirement_general)
        )

    return ""


experience_test = pd.DataFrame([
    {
        "id": job.get("id"),
        "title": job.get("title"),
        "experience": extract_toss_experience(job),
    }
    for job in target_toss_jobs
])

print("===== Toss experience 최종 추출 결과 =====")
print(
    "experience 있음:",
    (experience_test["experience"] != "").sum(),
    "/",
    len(experience_test)
)

display(
    experience_test[
        experience_test["experience"] != ""
    ].head(30)
)
```

```python
def is_sentence_like_candidate(text):
    text = str(text).strip()

    # 너무 긴 문자열은 설명 문장일 가능성이 높음
    if len(text) > 50:
        return True

    # 문장 종결 표현
    sentence_endings = (
        "해요",
        "있어요",
        "합니다",
        "됩니다",
        "이에요",
        "예요",
        "주세요",
        "좋아요",
        "있습니다",
    )

    if text.endswith(sentence_endings):
        return True

    # 한국어 문장형 표현
    sentence_keywords = [
        "활용하고",
        "사용하고",
        "지원하고",
        "시도하고",
        "구성원이",
        "생산성",
        "개발 도구를",
    ]

    if any(
        keyword in text
        for keyword in sentence_keywords
    ):
        return True

    return False


tech_candidate_df["sentence_like"] = (
    tech_candidate_df["candidate"]
    .apply(is_sentence_like_candidate)
)

print("===== Toss 후보 1차 분리 =====")

print(
    "기술명 후보:",
    (~tech_candidate_df["sentence_like"]).sum()
)

print(
    "문장형 후보:",
    tech_candidate_df["sentence_like"].sum()
)

print("\n===== 문장형 후보 샘플 =====")

display(
    tech_candidate_df[
        tech_candidate_df["sentence_like"]
    ].head(30)
)

print("\n===== 기술명 후보 샘플 =====")

display(
    tech_candidate_df[
        ~tech_candidate_df["sentence_like"]
    ].head(80)
)
```

```python
def is_valid_skill_name_candidate(text):
    text = str(text).strip()

    # 너무 긴 값은 기술명 후보에서 제외
    if len(text) > 45:
        return False

    # 문장형 표현 제외
    if is_sentence_like_candidate(text):
        return False

    # 문장에 자주 나타나는 조사/서술 표현 제외
    sentence_fragments = [
        "이해하고",
        "사용하고",
        "활용하고",
        "운영하며",
        "시도하고",
        "개발하고",
        "구축하고",
        "있어요",
        "합니다",
        "됩니다",
    ]

    if any(
        fragment in text
        for fragment in sentence_fragments
    ):
        return False

    return True


valid_candidates = [
    candidate
    for candidate in cleaned_candidate_frequency_df["candidate"]
    if is_valid_skill_name_candidate(candidate)
]


possible_duplicates_v2 = []

for i, a in enumerate(valid_candidates):
    a_norm = re.sub(
        r"[^a-z0-9]",
        "",
        a.lower()
    )

    for b in valid_candidates[i + 1:]:
        b_norm = re.sub(
            r"[^a-z0-9]",
            "",
            b.lower()
        )

        if (
            a_norm
            and a_norm == b_norm
            and a != b
        ):
            possible_duplicates_v2.append(
                (a, b)
            )


print("===== 표기 차이 후보 v2 =====")

for a, b in possible_duplicates_v2:
    print(
        f"{a}  <->  {b}"
    )
```

```python
# --------------------------------------------------
# Toss Skill Dictionary 기반 전체 공고 기술 추출
# --------------------------------------------------

# 최종 표준 기술명 집합
TOSS_STANDARD_SKILLS = set(
    expanded_skill_counter.keys()
)


# alias -> 표준 기술명 리스트
TOSS_SKILL_ALIAS_MAP = {}

# 1. 표준 기술명 자체도 alias로 등록
for skill in TOSS_STANDARD_SKILLS:
    TOSS_SKILL_ALIAS_MAP[skill] = [skill]


# 2. 표기 차이 normalization alias 등록
for alias, normalized in SKILL_NORMALIZATION_MAP.items():
    if normalized in COMPOSITE_SKILL_MAP:
        TOSS_SKILL_ALIAS_MAP[alias] = (
            COMPOSITE_SKILL_MAP[normalized]
        )
    else:
        TOSS_SKILL_ALIAS_MAP[alias] = [normalized]


# 3. 복합 원문 표현 등록
for alias, skills in COMPOSITE_SKILL_MAP.items():
    TOSS_SKILL_ALIAS_MAP[alias] = skills


def contains_skill_alias(text, alias):
    """
    짧은 기술명(C, Go, R 등)이 다른 단어 내부에서
    잘못 매칭되는 것을 막기 위한 경계 검사
    """
    pattern = (
        r"(?<![A-Za-z0-9])"
        + re.escape(alias)
        + r"(?![A-Za-z0-9])"
    )

    return re.search(
        pattern,
        text,
        flags=re.IGNORECASE
    ) is not None


def extract_toss_skills(job):
    description = get_toss_job_description(job)

    parsed = parse_toss_description(description)
    parsed = apply_toss_fallbacks(
        job,
        parsed
    )

    # 분석 대상:
    # 업무 + 지원자격 + 우대사항 + 명시적 기술스택
    skill_source_text = "\n".join([
        parsed.get("responsibilities", ""),
        parsed.get("requirements", ""),
        parsed.get("preferred", ""),
        parsed.get("tech_stack", ""),
    ])

    raw_matches = []
    std_matches = []

    # 긴 alias부터 검사
    aliases = sorted(
        TOSS_SKILL_ALIAS_MAP.keys(),
        key=len,
        reverse=True
    )

    for alias in aliases:
        if contains_skill_alias(
            skill_source_text,
            alias
        ):
            raw_matches.append(alias)

            for std_skill in (
                TOSS_SKILL_ALIAS_MAP[alias]
            ):
                std_matches.append(std_skill)

    # 순서 유지 + 중복 제거
    raw_matches = list(
        dict.fromkeys(raw_matches)
    )

    std_matches = list(
        dict.fromkeys(std_matches)
    )

    return raw_matches, std_matches


# --------------------------------------------------
# 156건 전체 테스트
# --------------------------------------------------

toss_skill_test = []

for job in target_toss_jobs:
    raw_skills, std_skills = extract_toss_skills(
        job
    )

    toss_skill_test.append({
        "id": job.get("id"),
        "title": job.get("title"),
        "category": map_toss_job_category(job),
        "raw_skill_count": len(raw_skills),
        "std_skill_count": len(std_skills),
        "skills_raw": " | ".join(raw_skills),
        "skills_std": " | ".join(std_skills),
    })


toss_skill_test_df = pd.DataFrame(
    toss_skill_test
)


print("===== Toss 156건 Skill Dictionary 적용 결과 =====")

print(
    "기술 1개 이상 추출:",
    (toss_skill_test_df["std_skill_count"] > 0).sum(),
    "/",
    len(toss_skill_test_df)
)

print(
    "기술 미추출:",
    (toss_skill_test_df["std_skill_count"] == 0).sum()
)

print(
    "평균 표준 기술 수:",
    round(
        toss_skill_test_df[
            "std_skill_count"
        ].mean(),
        2
    )
)

display(
    toss_skill_test_df.head(20)
)
```

```python
missing_skill_text_candidates = Counter()

for job in target_toss_jobs:
    raw_skills, std_skills = extract_toss_skills(job)

    if std_skills:
        continue

    description = get_toss_job_description(job)

    parsed = parse_toss_description(description)
    parsed = apply_toss_fallbacks(
        job,
        parsed
    )

    source_text = "\n".join([
        parsed.get("responsibilities", ""),
        parsed.get("requirements", ""),
        parsed.get("preferred", ""),
        parsed.get("tech_stack", ""),
    ])

    # 영문 기술명 후보 추출
    candidates = re.findall(
        r"""
        \b
        [A-Za-z][A-Za-z0-9.+/#_-]*
        (?:\s+[A-Za-z0-9.+/#_-]+){0,3}
        \b
        """,
        source_text,
        flags=re.VERBOSE
    )

    for candidate in candidates:
        candidate = candidate.strip()

        # 너무 짧거나 일반 문장 단어 제외
        if len(candidate) < 2:
            continue

        missing_skill_text_candidates[
            candidate
        ] += 1


missing_candidate_df = pd.DataFrame(
    missing_skill_text_candidates.items(),
    columns=[
        "candidate",
        "count",
    ]
).sort_values(
    "count",
    ascending=False
).reset_index(drop=True)


print("===== 기술 미추출 21건 영문 후보 =====")
print(
    "unique candidates:",
    len(missing_candidate_df)
)

display(
    missing_candidate_df.head(150)
)
```

```python
from datetime import datetime, timezone

TOSS_COLLECTED_AT = datetime.now(
    timezone.utc
).isoformat()

def build_toss_row(job):
    description = get_toss_job_description(job)

    parsed = parse_toss_description(description)
    parsed = apply_toss_fallbacks(
        job,
        parsed
    )

    raw_skills, std_skills = extract_toss_skills(
        job
    )

    raw_category = get_metadata_value(
        job,
        "커리어 페이지 노출 Job Category 값을 선택해주세요"
    )

    subsidiary = get_metadata_value(
        job,
        "포지션의 소속 자회사를 선택해 주세요."
    )

    employment_type = get_metadata_value(
        job,
        "Employment_Type"
    )

    location_data = job.get("location") or {}

    first_published = job.get(
        "first_published",
        ""
    )

    posted_date = ""

    if first_published:
        posted_date = (
            datetime
            .fromisoformat(first_published)
            .date()
            .isoformat()
        )

    source_job_id = str(
        job.get("id", "")
    )

    return {
        "job_id":
            f"toss_{source_job_id}",

        "source_job_id":
            source_job_id,

        "company":
            subsidiary or "",

        "title":
            job.get("title", ""),

        "job_category_raw":
            raw_category or "",

        "job_category_std":
            map_toss_job_category(job),

        "experience":
            extract_toss_experience(job),

        "employment_type":
            employment_type or "",

        "location":
            location_data.get(
                "name",
                ""
            ),

        "responsibilities":
            parsed.get(
                "responsibilities",
                ""
            ),

        "requirements":
            parsed.get(
                "requirements",
                ""
            ),

        "preferred":
            parsed.get(
                "preferred",
                ""
            ),

        "skills_raw":
            " | ".join(
                raw_skills
            ),

        "skills_std":
            " | ".join(
                std_skills
            ),

        "description_full":
            description,

        "posted_date":
            posted_date,

        # Toss 156건 모두 실제 마감일 값 없음
        "deadline":
            "",

        "source":
            "Toss Careers",

        "url":
            job.get(
                "absolute_url",
                ""
            ),

        "collected_at":
            TOSS_COLLECTED_AT,
    }
```
